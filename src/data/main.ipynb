```python
from pathlib import Path
import random
import shutil

VAL_RATIO = 0.10
SEED = 42

DATASET_DIR = Path("/home/jordan/Documents/WEBSITE/DermSight/src/data")
TRAIN_DIR = DATASET_DIR / "train"
VAL_DIR = DATASET_DIR / "val"

IMAGE_EXTENSIONS = {
    ".jpg", ".jpeg", ".png", ".bmp", ".webp"
}

random.seed(SEED)

for class_dir in sorted(TRAIN_DIR.iterdir()):
    if not class_dir.is_dir():
        continue

    images = [
        f for f in class_dir.iterdir()
        if f.is_file() and f.suffix.lower() in IMAGE_EXTENSIONS
    ]

    print(class_dir.name, len(images))

for class_dir in sorted(TRAIN_DIR.iterdir()):
    if not class_dir.is_dir():
        continue

    class_name = class_dir.name

    images = [
        f for f in class_dir.iterdir()
        if f.is_file() and f.suffix.lower() in IMAGE_EXTENSIONS
    ]

    random.shuffle(images)

    total = len(images)

    if total <= 1:
        continue

    val_count = max(1, round(total * VAL_RATIO))
    val_count = min(val_count, total - 1)

    val_images = images[:val_count]

    val_class_dir = VAL_DIR / class_name
    val_class_dir.mkdir(parents=True, exist_ok=True)

    for image_path in val_images:
        shutil.move(
            str(image_path),
            str(val_class_dir / image_path.name)
        )

    print(
        f"{class_name}: "
        f"train={total - val_count}, "
        f"val={val_count}"
    )

train_total = 0
val_total = 0

for class_dir in sorted(TRAIN_DIR.iterdir()):
    if not class_dir.is_dir():
        continue

    train_count = len([
        f for f in class_dir.iterdir()
        if f.is_file() and f.suffix.lower() in IMAGE_EXTENSIONS
    ])

    val_class_dir = VAL_DIR / class_dir.name

    val_count = 0

    if val_class_dir.exists():
        val_count = len([
            f for f in val_class_dir.iterdir()
            if f.is_file() and f.suffix.lower() in IMAGE_EXTENSIONS
        ])

    train_total += train_count
    val_total += val_count

    print(
        f"{class_dir.name:<25} "
        f"train={train_count:<5} "
        f"val={val_count:<5}"
    )

print("\nTotal train:", train_total)
print("Total val:", val_total)

print(
    "Validation ratio:",
    val_total / (train_total + val_total)
)
```
